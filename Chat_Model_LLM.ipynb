```python
import langchain #the main library to use langchain components.
from langchain_google_genai import GoogleGenerativeAI, ChatGoogleGenerativeAI # it is langchain package to connect with gemini models through langchain
# (class-1 : provided by langchain): GoogleGenerativeAI: it is a model componuent used for regular LLM to access Gemini model
# (class-2 : provided by langchain):ChatGoogleGenerativeAI:  it is a model componuent act as a wrapper to access chat model()LLM and access Gemini model
```

```python
#Access your secret keys in Python via:
from google.colab import userdata
gen =userdata.get('gemini_lite')
```

```python
import os

os.environ["GOOGLE_API_KEY"] = gen

#We have to set up the environment variable to access gemini model

#os.environ is a dictionary provided by Python that stores environment variables for the current program.
#Think of environment variables as: 🔐 secure global configuration values that programs can read without hard-coding them.


#API key is used for: Authentication, Security (Google must know who you are)

#The name of the environment variable must be exactly: "GOOGLE_API_KEY"
#Because Google Gemini SDK & LangChain internally look for this exact key name.
#If the name is different → authentication fails.

# Same rule applies to other models
# 1. OpenAI -> OPENAI_API_KEY
# 2. Google Gemini -> GOOGLE_API_KEY
# 3. Anthropic -> ANTHROPIC_API_KEY
# 4. HuggingFace -> HUGGINGFACEHUB_API_TOKEN
# 5. Cohere -> COHERE_API_KEY
```

```python
os.environ.pop("GEMINI_API_KEY", None)

# It safely removes an environment variable if it exists, preventing conflicts or authentication issues, without raising an error if the variable is absent.
```

```python
#All the gemini model as chat model for having a conversation
```

```python

cmodel = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
message = [("system","you area 9 months old child"),
           ("user","Tell me about apple")]
```

```python
#invocation:
response = cmodel.invoke(message)

# Why use invoke(message) -> Because we want one response like normal question - Answers
# " Ask once -> get full answer "
```

```python
response # See in Output, There is "content=" so we just need to access that part only
```

### A simple chatbot with langchain

```python
cmodel = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
message = [("system","you are 12 year old child"),
           ("user","Hi my name is shrutika")]

response = cmodel.invoke(message)
response.content
```

```python
# I take the same model:

message = [("system","you are 12 year old child"),
           ("user","Hi what is my name")]
response = cmodel.invoke(message)
response.content

#We store chat memory by appending the AI’s responses to the message history so future interactions remain context-aware.
```

```python
brain = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
message=[("system","You are having a 10 plus years of experience in python programming")]

while True:
  user = input("Type a user message :")

  message.append(("user",user))

  ai =brain.invoke(message)

  message.append(("ai",ai.content))

  print("ai:{}".format(ai.content))

  if user.lower() =="bye":
    break
```
