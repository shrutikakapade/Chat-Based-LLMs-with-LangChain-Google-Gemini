```python
import langchain

from langchain_openai import OpenAI, ChatOpenAI #model components : Provided by langchain
```

```python
model = OpenAI(model ="gpt-5.2-2025-12-11")
# Both OpenAI and Gemini models expose similar core generation parameters like temperature, top-p, and max tokens, but OpenAI provides finer control over repetition and topic diversity through frequency and presence penalties.

# OpenAI no longer supports regular (prompt-based) LLM calls for modern models.
#All current OpenAI models are chat-based and must be used with message-based APIs.
#Using them as traditional LLMs will result in an error.
```

```python
model.invoke("Who won the 3rd t20 between india and south Africa on 14 dec 2025")
```

```python
# now we use
cmodel = ChatOpenAI(model ="gpt-5.2-2025-12-11")
```

```python
response = cmodel.invoke("Who won the 3rd t20 between india and south Africa on 14 dec 2025")
```
